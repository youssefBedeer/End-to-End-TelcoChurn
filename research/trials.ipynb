```python
import urllib.request as request
from src import logging, CustomException
import os 


def download_file():
    if not os.path.exists("artifacts/data_ingestion/data.zip"):
        try:
            filename, header = request.urlretrieve(
                url = "https://github.com/youssefBedeer/data-repository/raw/refs/heads/main/Telco-Customer-Churn.rar",
                filename= "artifacts/data_ingestion/data.zip"
            )
            logging.info(f"{filename} download! with the following info: \n{header}")
        except Exception as e:
            raise CustomException(e) from e
        
    else:
        logging.info(f"File '{filename}' already exist")
        
```

```python
import pandas as pd 
df = pd.read_csv("artifacts/data_ingestion/Telco-Customer-Churn.csv")
df.head(1)
```

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
object_cols = df.select_dtypes(include=['object']).columns.tolist()
```

```python
numerical_cols
```

```python
object_cols
```

```python
converted_cols = []

for col in object_cols:
    try:
        # Try converting; use 'coerce' to handle spaces or bad values gracefully
        temp = pd.to_numeric(df[col], errors="coerce")

        # If the conversion produces mostly numbers (not all NaN), keep it
        if temp.notna().mean() > 0.5:  # adjustable threshold
            df[col] = temp
            converted_cols.append(col)
    except Exception as e:
        print(f"Could not convert {col}: {e}")

print("Converted columns:", converted_cols)
```

```python
df = pd.DataFrame({
    'age': ['23', '45', '30'],
    'height': ['175.5', '180.2', '169.8'],
    'name': ['Alice', 'Bob', 'Charlie']
})

for col in df.columns:
    if df[col].dtype == 'object':
        try:
            df[col] = pd.to_numeric(df[col], errors='raise')
        except:
            pass

print(df.dtypes)
```
